We continue with `customer_churn.csv` from the lecture (7,043 customers, 26.5% churners).

*Demo note: the CSV in this folder is a synthetic stand-in with the same shape as the lecture data.*

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv('customer_churn.csv')
df.head()
```

**Question:** Compute the churn rate, i.e. the fraction of customers with `Churn == 'Yes'`, and store it in a variable `churn_rate` (a float between 0 and 1).

```python
### BEGIN SOLUTION
churn_rate = (df['Churn'] == 'Yes').mean()
### END SOLUTION
```

```python
assert 0 <= churn_rate <= 1, 'churn_rate must be a fraction'
assert abs(churn_rate - 0.265) < 0.005

### BEGIN HIDDEN TESTS
assert isinstance(churn_rate, float)
### END HIDDEN TESTS
```

**Question:** Split the data with `train_test_split(..., test_size=0.25, random_state=0, stratify=y)` using the features `tenure` and `MonthlyCharges`. Train a logistic regression and store the **recall for the churn class** on the test set in `recall`.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import recall_score

X = df[['tenure', 'MonthlyCharges']]
y = (df['Churn'] == 'Yes').astype(int)
### BEGIN SOLUTION
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0, stratify=y)
model = LogisticRegression().fit(X_train, y_train)
recall = recall_score(y_test, model.predict(X_test))
### END SOLUTION
print(f'Recall: {recall:.3f}')
```

```python
assert 0 <= recall <= 1, 'recall must be between 0 and 1'

### BEGIN HIDDEN TESTS
from sklearn.linear_model import LogisticRegression as _LR
from sklearn.metrics import recall_score as _recall
_y = (df['Churn'] == 'Yes').astype(int)
_X_train, _X_test, _y_train, _y_test = train_test_split(df[['tenure', 'MonthlyCharges']], _y, test_size=0.25, random_state=0, stratify=_y)
assert abs(recall - _recall(_y_test, _LR().fit(_X_train, _y_train).predict(_X_test))) < 1e-6, 'recall differs from the expected value'
### END HIDDEN TESTS
```

**Question:** Plot the distribution of `MonthlyCharges` for churners and non-churners in one figure, so the two groups can be compared.

```python
import matplotlib.pyplot as plt

### BEGIN SOLUTION
for label, group in df.groupby('Churn'):
    plt.hist(group['MonthlyCharges'], bins=30, alpha=0.5, density=True, label=f'Churn = {label}')
plt.xlabel('Monthly charges')
plt.ylabel('Density')
plt.legend()
plt.show()
### END SOLUTION
```
